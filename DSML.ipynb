Missing values were checked using `isnull().sum()`. Missing values in categorical columns were replaced with the most frequent value (mode). Duplicate rows were identified and removed using `drop_duplicates()`. Negative numerical values and invalid ages were checked to identify possible data errors. These methods help improve data quality before training the machine learning model. The columns containing missing values were **[write column names from the output]**.

```python

print("Missing Values Before Cleaning:")
display(df.isnull().sum())

categorical_cols = df.select_dtypes(
    include=['object', 'category']
).columns

for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("\nDuplicate Rows Before Cleaning:", df.duplicated().sum())
df = df.drop_duplicates()

print("\nNegative Values in Numeric Columns:")
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    negative_count = (df[col] < 0).sum()
    if negative_count > 0:
        print(f"{col}: {negative_count} negative values")

if 'age' in df.columns:
    print("\nInvalid Ages (age <= 0 or age > 120):",
          ((df['age'] <= 0) | (df['age'] > 120)).sum())

if 'income' in df.columns:
    print("Negative Incomes:", (df['income'] < 0).sum())

print("\nMissing Values After Cleaning:")
display(df.isnull().sum())

print("\nDataset Shape After Cleaning:", df.shape)
```

```python

df = df.drop(columns=['customer_id'])

df['loan_default'] = df['loan_default'].map({'Yes': 1, 'No': 0})

df = pd.get_dummies(df, drop_first=True, dtype=int)

X = df.drop('loan_default', axis=1)
y = df['loan_default']

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeature columns:")
print(X.columns.tolist())

print("\nTarget distribution:")
print(y.value_counts())
```

We split the dataset into 80% training data and 20% testing data. The training data is used to train the model, while the testing data evaluates its performance on unseen data. Stratification maintains the proportion of defaulting and non-defaulting customers in both sets.

```python

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print("Training features shape:", X_train.shape)
print("Testing features shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```

```python

from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)

y_train_pred = dt.predict(X_train)
y_test_pred = dt.predict(X_test)

print("Training Accuracy:", dt.score(X_train, y_train))
print("Testing Accuracy:", dt.score(X_test, y_test))
```

The training accuracy is 100%, while the testing accuracy is approximately 76.12%. There is a significant gap between the two scores, suggesting that the Decision Tree is overfitting the training data and may not generalise well to unseen customers.


```python

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, classification_report, roc_auc_score

cm = confusion_matrix(y_test, y_test_pred)

disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap="Blues")
plt.show()

print(classification_report(y_test, y_test_pred, target_names=["No Default", "Default"]))

y_test_prob = dt.predict_proba(X_test)[:, 1]
print("ROC-AUC Score:", roc_auc_score(y_test, y_test_prob))
```

Approving a loan to a customer who later defaults can cause financial losses for the bank. Therefore, recall for the Default class is important because it measures how many actual defaulters the model correctly identifies. However, rejecting a good customer is also undesirable, so the bank should balance recall with precision.

A maximum depth of 5 gave the best test accuracy among the four manually tested depths, at approximately 82.59%. GridSearchCV selected a maximum depth of 3, a minimum leaf size of 10, and the gini criterion, achieving a test accuracy of 81.59%. Although tuning reduced the gap between training and testing performance, the tuned model has very low recall for defaulters, so accuracy alone is not sufficient to judge its usefulness.

```python

from sklearn.tree import DecisionTreeClassifier

for depth in [3, 5, 7, 10]:
    model = DecisionTreeClassifier(max_depth=depth, random_state=42)
    model.fit(X_train, y_train)

    print("Max Depth:", depth)
    print("Training Accuracy:", model.score(X_train, y_train))
    print("Testing Accuracy:", model.score(X_test, y_test))
    print()
```

I would recommend Logistic Regression as the initial model because it achieved the highest accuracy of 83.6%, recall of 27.0%, F1-score of 0.377, and ROC-AUC of 0.760 among the three models. However, its recall is still low, meaning it misses many actual defaulters. The bank should consider recall and the cost of incorrect predictions before selecting the final model.

```python

from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

param_grid = {
    'max_depth': [3, 5, 7, 10, None],
    'min_samples_leaf': [1, 2, 5, 10],
    'criterion': ['gini', 'entropy']
}

grid = GridSearchCV(
    DecisionTreeClassifier(random_state=42),
    param_grid,
    cv=5,
    scoring='accuracy'
)

grid.fit(X_train, y_train)

best = grid.best_estimator_

print("Best Parameters:", grid.best_params_)
print("Best Cross-Validation Accuracy:", grid.best_score_)
print("Test Accuracy:", accuracy_score(y_test, best.predict(X_test)))
print(classification_report(y_test, best.predict(X_test)))
```

```python

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, recall_score, f1_score, roc_auc_score
import pandas as pd

lr = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000, random_state=42)
)

lr.fit(X_train, y_train)

models = {
    "Logistic Regression": lr,
    "Untuned Decision Tree": dt,
    "Tuned Decision Tree": best
}

results = []

for name, model in models.items():
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1 Score": f1_score(y_test, y_pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    })

results_df = pd.DataFrame(results)
print(results_df.round(3))
```
